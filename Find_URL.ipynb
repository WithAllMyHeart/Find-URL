```python

import re

def find_gov_urls(website):
 pattern = r'([a-z]{5}\:\/+[a-z]{3}\.\w+\w+\.[a-z]{3})'
 result = re.findall(pattern, website)
 return result

print(find_gov_urls("https://www.data.gov is a great place to find open source datasets!")) # Should return ['https://www.data.gov']

print(find_gov_urls("Learn more about US National Parks at https://www.nps.gov, https://www.nationalparks.org, or https://www.recreation.gov.")) # Should return ['https://www.nps.gov', 'https://www.nationalparks.org',  'https://www.recreation.gov']

print(find_gov_urls("The Library of Congress (https://www.loc.gov) is an incredible resource!")) # Should return ['https://www.loc.gov']
```
